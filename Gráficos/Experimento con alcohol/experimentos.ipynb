```python
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# CSV FILE
# ============================================================

archivo = "mediciones/medición_etanol.csv"   # Change this to your CSV filename


# ============================================================
# READ REFERENCE FREQUENCY FROM HEADER
# ============================================================

f_ref = None

with open(archivo, "r", encoding="utf-8") as f:
    for linea in f:
        if linea.startswith("# f0_real="):
            f_ref = float(
                linea.split("=")[1]
                     .replace("Hz", "")
                     .strip()
            )
            break

if f_ref is not None:
    print(f"Frecuencia de referencia: {f_ref:.3f} Hz")
    print(f"Frecuencia de referencia: {f_ref/1e6:.6f} MHz")
else:
    print("No se encontró la frecuencia de referencia en el encabezado.")


# ============================================================
# READ DATA
# ============================================================

# Lines beginning with # are ignored
df = pd.read_csv(archivo, comment="#")

display(df.head())


# ============================================================
# PLOT DELTA F
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(df["timestamp_s"], df["delta_f_hz"])

plt.xlabel("Tiempo [s]")
plt.ylabel("Δf [Hz]")
plt.title("Variación de frecuencia")

plt.grid(True)
plt.tight_layout()
plt.show()


# ============================================================
# PLOT THICKNESS
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(df["timestamp_s"], df["espesor_nm"])

plt.xlabel("Tiempo [s]")
plt.ylabel("Espesor [nm]")
plt.title("Espesor calculado")

plt.grid(True)
plt.tight_layout()
plt.show()
```

# **2 SENSORES**

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = "mediciones/2_sensores_caja.csv"

df = pd.read_csv(file_path, comment="#")


# ============================================================
# Check ranges
# ============================================================

print("Sensor frequency range:")
print(df["f_sensor_hz"].min(), "to", df["f_sensor_hz"].max(), "Hz")

print("\nReference frequency range:")
print(df["f_ref_hz"].min(), "to", df["f_ref_hz"].max(), "Hz")

print("\nDelta f range:")
print(df["delta_f_hz"].min(), "to", df["delta_f_hz"].max(), "Hz")


# ============================================================
# PLOT 1 — Delta f
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    df["timestamp_s"],
    df["delta_f_hz"],
    linewidth=1.2
)

plt.axhline(0, linestyle="--", linewidth=0.8)

plt.xlabel("Time [s]")
plt.ylabel(r"$\Delta f$ [Hz]")
plt.title("Frequency Difference Between Crystals")

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# PLOT 2 — Sensor and reference crystal frequencies
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    df["timestamp_s"],
    df["f_sensor_hz"],
    label="Sensor crystal",
    linewidth=1.2
)

plt.plot(
    df["timestamp_s"],
    df["f_ref_hz"],
    label="Reference crystal",
    linewidth=1.2
)

plt.xlabel("Time [s]")
plt.ylabel("Frequency [Hz]")
plt.title("Sensor and Reference Crystal Frequencies")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
