# **Primeras prueba del cristal de cuarzo sin encapsulado**

La prueba consiste en hacer oscilar el cristal mediante 1 minuto e ir tomando muestras cada 1 segundo. Al principio se deja oscilar libremente y luego se sopla el cristal para sensar cuanto varia de su frecuencia en reposo

### **Primera medición**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Nombre del archivo
archivo = "archivos/medicion.txt"

# Leer el archivo
datos = pd.read_csv(archivo, sep="\t")

# Mostrar los primeros datos para verificar
print(datos.head())

# Graficar
plt.figure(figsize=(10, 5))

plt.plot(
    datos["tiempo_s"],
    datos["frecuencia_hz"],
    marker="o",
    markersize=3,
    linewidth=1
)

plt.xlabel("Tiempo [s]")
plt.ylabel("Frecuencia [Hz]")
plt.title("Frecuencia del cristal en función del tiempo")

plt.grid(True)
plt.tight_layout()

plt.show()
```
